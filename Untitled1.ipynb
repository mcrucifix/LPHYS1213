```python
import numpy as np
import matplotlib.pyplot as plt

# --- Parameters ---
N = 201           # grid points (odd is convenient)
L = 2.5           # half-width of square domain: x,y in [-L,L]
U = 1.0           # freestream speed (phi -> U x at outer boundary)
R = 0.6           # cylinder radius
omega = 1.7       # SOR parameter (1 < omega < 2)
tol = 1e-5
max_iter = 20000

# --- Grid ---
x = np.linspace(-L, L, N)
y = np.linspace(-L, L, N)
dx = x[1] - x[0]
X, Y = np.meshgrid(x, y, indexing='ij')

# cylinder mask: True inside obstacle
cyl = (X**2 + Y**2) <= R**2

# fluid mask
fluid = ~cyl

# --- Initial phi: freestream potential phi = U*x ---
phi = U * X.copy()

# Helper: indices of interior fluid points (exclude outer boundary)
I, J = np.arange(1, N-1), np.arange(1, N-1)

# --- SOR solver with discrete Neumann on cylinder ---
for it in range(max_iter):
    maxdiff = 0.0
    for i in I:
        for j in J:
            if not fluid[i, j]:
                continue  # skip points inside cylinder
            # outer boundary kept as Dirichlet (freestream)
            if i == 0 or i == N-1 or j == 0 or j == N-1:
                phi[i, j] = U * x[i]
                continue

            # Laplace update with Neumann for neighbours that are inside cylinder:
            s = 0.0
            # neighbour (i+1,j)
            if fluid[i+1, j]:
                s += phi[i+1, j]
            else:
                # neighbour inside cylinder -> use mirrored value (phi_ghost = phi[i,j])
                s += phi[i, j]
            # neighbour (i-1,j)
            if fluid[i-1, j]:
                s += phi[i-1, j]
            else:
                s += phi[i, j]
            # neighbour (i,j+1)
            if fluid[i, j+1]:
                s += phi[i, j+1]
            else:
                s += phi[i, j]
            # neighbour (i,j-1)
            if fluid[i, j-1]:
                s += phi[i, j-1]
            else:
                s += phi[i, j]

            phi_new = 0.25 * s
            # SOR relax
            phi_old = phi[i, j]
            phi[i, j] = phi_old + omega * (phi_new - phi_old)
            maxdiff = max(maxdiff, abs(phi[i, j] - phi_old))
    if it % 200 == 0:
        print(f"iter {it}, maxΔ = {maxdiff:.3e}")
    if maxdiff < tol:
        print(f"Converged after {it} iterations (maxΔ={maxdiff:.2e})")
        break
else:
    print("Reached max_iter without full convergence; maxΔ =", maxdiff)

# --- Compute velocities by central differences (only where neighbours are fluid) ---
u = np.zeros_like(phi)
v = np.zeros_like(phi)
for i in range(1, N-1):
    for j in range(1, N-1):
        # central differences; if neighbour inside cylinder, fall back to one-sided
        if fluid[i+1,j] and fluid[i-1,j]:
            u[i,j] = (phi[i+1,j] - phi[i-1,j]) / (2*dx)
        elif fluid[i+1,j]:
            u[i,j] = (phi[i+1,j] - phi[i,j]) / dx
        elif fluid[i-1,j]:
            u[i,j] = (phi[i,j] - phi[i-1,j]) / dx
        else:
            u[i,j] = 0.0

        if fluid[i,j+1] and fluid[i,j-1]:
            v[i,j] = (phi[i,j+1] - phi[i,j-1]) / (2*dx)
        elif fluid[i,j+1]:
            v[i,j] = (phi[i,j+1] - phi[i,j]) / dx
        elif fluid[i,j-1]:
            v[i,j] = (phi[i,j] - phi[i,j-1]) / dx
        else:
            v[i,j] = 0.0

speed = np.sqrt(u**2 + v**2)

# --- Streamfunction psi by integrating ∂ψ/∂x = -v (choose psi=0 at left boundary) ---
psi = np.zeros_like(phi)
for i in range(N-1):
    # integrate from x[i] to x[i+1] for all y (trapezoidal)
    psi[i+1,:] = psi[i,:] - 0.5*(v[i+1,:] + v[i,:]) * dx

# Mask psi inside cylinder for plotting clarity
psi_masked = np.ma.array(psi, mask=~fluid)

# --- Locate approximate stagnation points near cylinder: points on cylinder boundary with small speed ---
# find grid points adjacent to cylinder (fluid points with at least one neighbour inside)
adjacent = np.zeros_like(fluid, dtype=bool)
for i in range(1,N-1):
    for j in range(1,N-1):
        if not fluid[i,j]:
            continue
        if (not fluid[i+1,j]) or (not fluid[i-1,j]) or (not fluid[i,j+1]) or (not fluid[i,j-1]):
            adjacent[i,j] = True

cand = np.where(adjacent & (speed < 0.05*U))  # threshold
stagn_points = [(x[i], y[j]) for i,j in zip(cand[0], cand[1])]
print("Approximate stagnation points (x,y):", stagn_points)

# --- Plot results ---
fig, ax = plt.subplots(1,2, figsize=(12,5))
ax[0].contour(X, Y, phi, levels=40)
ax[0].add_patch(plt.Circle((0,0), R, color='k', fill=False))
ax[0].set_title('Potential φ (contours)')
ax[0].set_aspect('equal')

cs = ax[1].contour(X, Y, psi_masked, levels=40)
ax[1].add_patch(plt.Circle((0,0), R, color='k', fill=False))
ax[1].set_title('Streamfunction ψ (contours)')
ax[1].set_aspect('equal')

# mark stagnation points if any
for sx, sy in stagn_points:
    ax[0].plot(sx, sy, 'ro')
    ax[1].plot(sx, sy, 'ro')

plt.show()
```
