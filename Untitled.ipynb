```julia
using Gridap
using GridapGmsh

# ----------------------------------------------------------
# 1. Build annulus mesh in Gmsh
# ----------------------------------------------------------
import Gmsh: gmsh

gmsh.initialize()
gmsh.model.add("annulus")

R  = 1.0
R2 = 5.0

# Circles
c1 = gmsh.model.occ.addCircle(0,0,0,R)
c2 = gmsh.model.occ.addCircle(0,0,0,R2)

# Plane surface with hole
ll  = gmsh.model.occ.addCurveLoop([c2])
lh  = gmsh.model.occ.addCurveLoop([c1])
surf = gmsh.model.occ.addPlaneSurface([ll, lh])

gmsh.model.occ.synchronize()

# Physical groups (boundary tags)
outer = gmsh.model.addPhysicalGroup(1,[c2])
gmsh.model.setPhysicalName(1,outer,"outer")

inner = gmsh.model.addPhysicalGroup(1,[c1])
gmsh.model.setPhysicalName(1,inner,"inner")

# Mesh
gmsh.model.mesh.generate(2)
gmsh.write("annulus.msh")
gmsh.finalize()

# ----------------------------------------------------------
# 2. Load mesh into Gridap
# ----------------------------------------------------------
model = GmshDiscreteModel("annulus.msh")

Ω  = Triangulation(model)
Γo = BoundaryTriangulation(model, tags="outer")
Γi = BoundaryTriangulation(model, tags="inner")

# ----------------------------------------------------------
# 3. FE spaces
# ----------------------------------------------------------
order = 1
reffe = ReferenceFE(lagrangian, Float64, order)

V = TestFESpace(model, reffe; conformity=:H1)
U = TrialFESpace(model, reffe; conformity=:H1)

# ----------------------------------------------------------
# 4. Weak form of Laplace equation
# ----------------------------------------------------------
a(u,v) = ∫( ∇(u)⋅∇(v) )*dΩ

# Uniform flow potential φ = U∞ x
U∞ = 1.0
x = get_coordinates(Ω)
phi_outer = U∞ * x[1]

# Neumann on inner boundary (no penetration): natural → nothing to impose

l(v) = ∫(phi_outer * v)*dΓo   # outer Dirichlet via weak imposition

op = AffineFEOperator(a,l,U,V)

φh = solve(op)

writevtk(Ω,"laplace_annulus",cellfields=["phi"=>φh])

```
