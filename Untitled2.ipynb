```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
U = 1.0          # free-stream velocity
R = 1.0          # cylinder radius
N = 400          # grid resolution

# Cartesian grid
x = np.linspace(-3*R, 3*R, N)
y = np.linspace(-3*R, 3*R, N)
X, Y = np.meshgrid(x, y)

# Convert to polar coordinates
Rgrid = np.sqrt(X**2 + Y**2)
Theta = np.arctan2(Y, X)

# Streamfunction for flow past a cylinder
# psi = U (r - R^2/r) sin(theta)
psi = U * (Rgrid - R**2 / Rgrid) * np.sin(Theta)

# Mask inside the cylinder (r < R)
psi = np.ma.array(psi, mask=(Rgrid < R))

# Plot
plt.figure(figsize=(6,6))
levels = np.linspace(-3, 3, 40)
plt.contour(X, Y, psi, levels=levels)
circle = plt.Circle((0,0), R, color='black', fill=False, linewidth=2)
plt.gca().add_artist(circle)
plt.gca().set_aspect('equal', 'box')

plt.xlabel("x")
plt.ylabel("y")
plt.title("Streamfunction around a circular cylinder")
plt.tight_layout()
plt.show()
```
